```python
import os
# 모든 단계는 저장소 루트의 data/ 아래를 공유한다 (노트북은 cwd가 자신의 폴더라고 가정).
REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), "..", ".."))
CORPUS_DATA = os.path.join(REPO_ROOT, "data", "1.corpus")

import os
from tqdm import tqdm

def merge_datasets():
    # 입력 파일 경로 설정
    input_files = [
        os.path.join(CORPUS_DATA, "1.wiki/cleaned/kowiki_dataset_cleaned.txt"),
        os.path.join(CORPUS_DATA, "1.wiki/cleaned/namuwiki_cleaned.txt")
    ]
    
    # 출력 파일 경로 설정
    output_file = os.path.join(CORPUS_DATA, "1.wiki/outputs/merged_wiki.txt")
    
    # 출력 디렉토리가 없다면 생성
    os.makedirs(os.path.dirname(output_file), exist_ok=True)
    
    print(f"🚀 Starting to merge {len(input_files)} files into {output_file}...")
    
    # 출력 파일 쓰기 모드로 열기
    with open(output_file, 'w', encoding='utf-8') as outfile:
        for file_path in input_files:
            if not os.path.exists(file_path):
                print(f"❌ Error: '{file_path}' not found. Skipping.")
                continue
            
            # 파일 크기 계산 (진행률 표시용)
            file_size = os.path.getsize(file_path)
            print(f"\n📄 Processing: {file_path} ({(file_size / (1024*1024*1024)):.2f} GB)")
            
            # 대용량 파일을 메모리 안전하게 한 줄씩 읽어서 쓰기
            with open(file_path, 'r', encoding='utf-8') as infile:
                with tqdm(total=file_size, unit='B', unit_scale=True, desc="Merging") as pbar:
                    for line in infile:
                        outfile.write(line)
                        # UTF-8 바이트 기준으로 진행률 업데이트
                        pbar.update(len(line.encode('utf-8')))
            
            # 문서 간 구분을 확실히 하기 위해 파일 사이에 줄바꿈 추가
            outfile.write("\n\n")

    print("\n" + "="*60)
    print("✅ Merging Completed Successfully!")
    print(f"📂 Final Output File: {os.path.abspath(output_file)}")
    print("="*60)
```

```python
merge_datasets()
```
